# Pregunta 1 - Best Creatives

Ranking robusto desde `dataset/final/creative_cbr_cases_final.parquet`. El CBR se usa solo como capa explicativa de los winners.


```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet').exists():
    ROOT = ROOT.parent

import sys
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from creative_intelligence.cbr import CreativeMemory, load_cases
from creative_intelligence.questions import build_winner_ranking, write_question1_outputs

cases = load_cases(ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet')
memory = CreativeMemory(cases_path=ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet').fit()
ranking = build_winner_ranking(cases)
paths = write_question1_outputs(ranking, memory, ROOT / 'best_creatives' / 'outputs')
paths
```
