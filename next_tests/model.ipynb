# Pregunta 3 - What Should We Test Next?

Usa el CBR como memoria creativa: compara el creative con vecinos similares y propone el siguiente test.


```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet').exists():
    ROOT = ROOT.parent

import sys
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from creative_intelligence.cbr import CreativeMemory, load_cases
from creative_intelligence.questions import (
    build_winner_ranking,
    build_health_table,
    build_next_test_table,
    write_question3_outputs,
)

cases = load_cases(ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet')
memory = CreativeMemory(cases_path=ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet').fit()
ranking = build_winner_ranking(cases)
health = build_health_table(cases, memory, k=20)
next_tests, cards = build_next_test_table(cases, ranking, health, memory, k=20)
paths = write_question3_outputs(next_tests, cards, ROOT / 'next_tests' / 'outputs')
paths
```
