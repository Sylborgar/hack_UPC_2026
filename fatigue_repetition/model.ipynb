# Pregunta 2 - Tired & Repetitive Creatives

Detecta fatiga con senales de performance y repeticion con vecinos similares del modulo CBR.


```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet').exists():
    ROOT = ROOT.parent

import sys
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from creative_intelligence.cbr import CreativeMemory, load_cases
from creative_intelligence.questions import build_health_table, write_question2_outputs

cases = load_cases(ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet')
memory = CreativeMemory(cases_path=ROOT / 'dataset' / 'final' / 'creative_cbr_cases_final.parquet').fit()
health = build_health_table(cases, memory, k=20)
paths = write_question2_outputs(health, ROOT / 'fatigue_repetition' / 'outputs')
paths
```
